```python
%matplotlib inline
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ---------------------------
# 步骤1：生成数据
# ---------------------------
# 创建0到2π之间的100个等间隔点（x轴数据）
x = np.linspace(-1 * 2 * np.pi, 2 * np.pi, 400)

# 计算正弦和余弦值（y轴数据）
y_sin = np.sin(x)
y_cos = np.cos(x)
```

```python
# ---------------------------
# 步骤2：配置图表参数
# ---------------------------
# 设置图表尺寸（宽8英寸，高5英寸）
plt.figure(figsize=(8, 5))
```

```python
# ---------------------------
# 步骤3：绘制曲线
# ---------------------------
# 绘制正弦曲线（蓝色实线，标签用于图例）
plt.plot(
    x, y_sin, 
    color='blue',    # 颜色设为蓝色
    linestyle='-',   # 线型为实线
    linewidth=2,     # 线宽2像素
    label='sin(x)'   # 标签用于图例显示
)

# 绘制余弦曲线（红色虚线，标签用于图例）
plt.plot(
    x, y_cos, 
    color='red',     # 颜色设为红色
    linestyle='--',  # 线型为虚线
    linewidth=2,     # 线宽2像素
    label='cos(x)'   # 标签用于图例显示
)
```

```python
# ---------------------------
# 步骤4：添加图表元素
# ---------------------------
# 设置标题（字体大小14）
plt.title('Sine and Cosine Curves', fontsize=14)

# 设置x轴标签（弧度单位）
plt.xlabel('x (radians)', fontsize=12)

# 设置y轴标签
plt.ylabel('y', fontsize=12)

# 显示图例（自动选择最佳位置）
plt.legend(fontsize=12)

# 显示网格（浅灰色虚线）
plt.grid(linestyle='--', alpha=0.6)
```

```python
# ---------------------------
# 步骤5：显示图表
# ---------------------------
plt.show()    
```
