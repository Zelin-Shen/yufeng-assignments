```python
import pandas as pd
```

```python
class ExcelDataOrganizer:
    def __init__(self, input_file, output_file):
        self.input_file = input_file
        self.output_file = output_file
        self.df = None

    def load_data(self):
        """加载CSV数据，处理编码和年份格式"""
        # 使用utf-8-sig编码处理带BOM的CSV
        self.df = pd.read_csv(self.input_file, encoding='utf-8-sig')
        # 提取年份（处理“2012年”格式为2012）
        self.df['年份'] = self.df['年份'].str.extract(r'(\d+)').astype(int)

    def process_data(self):
        """按年份分组数据"""
        self.year_groups = self.df.groupby('年份')

    def write_to_excel(self):
        """将每个年份数据写入独立Sheet"""
        with pd.ExcelWriter(self.output_file, engine='openpyxl', mode='w') as writer:
            for year, group in self.year_groups:
                # 写入当前年份Sheet，自动处理列名
                group.to_excel(writer, sheet_name=str(year), index=False)

    def execute(self):
        """执行全流程"""
        self.load_data()
        self.process_data()
        self.write_to_excel()
```

```python
# 配置文件路径
input_csv = "ch10_input_data.csv"
output_excel = "按年份分组数据.xlsx"
```

```python
#实例化
organizer = ExcelDataOrganizer(input_csv, output_excel)
organizer.execute()
```

```python
print(f"数据已按年份整理到{output_excel}，每个年份对应独立Sheet")
```
