```python
class KthLargestFinder:
    def __init__(self):
        pass  # 初始化方法，可预留扩展接口

    def find_kth_largest(self, nums: list, k: int) -> int:
        """
        公共方法：查找第K大元素（1-based索引）
        :param nums: 输入数组
        :param k: 第K大（例如k=2表示第二大元素）
        :return: 第K大元素的值
        """
        return self._quick_select(nums, 0, len(nums) - 1, k)

    def _quick_select(self, nums: list, left: int, right: int, k: int) -> int:
        """
        私有方法：快速选择算法核心（降序分区查找第K大）
        :param nums: 输入数组（原地修改）
        :param left: 左边界索引
        :param right: 右边界索引
        :param k: 第K大（1-based）
        :return: 目标元素的值
        """
        pivot = nums[right]  # 选择右边界作为基准值
        i = left  # 记录基准值最终位置的前一个索引

        # 分区：将大于等于基准值的元素移到左侧
        for j in range(left, right):
            if nums[j] >= pivot:
                nums[i], nums[j] = nums[j], nums[i]
                i += 1

        # 基准值归位
        nums[i], nums[right] = nums[right], nums[i]

        # 目标索引为升序数组中的第(len(nums)-k)个位置（0-based）
        target_index = len(nums) - k
        if i == target_index:
            return nums[i]  # 找到目标
        elif i > target_index:  # 目标在左半区
            return self._quick_select(nums, left, i - 1, k)
        else:  # 目标在右半区
            return self._quick_select(nums, i + 1, right, k)
```
